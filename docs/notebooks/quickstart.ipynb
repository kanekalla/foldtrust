# FoldTrust Quickstart: SARS-CoV-2 FSE Analysis

This notebook demonstrates the scanpy-style FoldTrust API for RNA structure analysis.

```python
import sys
sys.path.insert(0, '../../src')

import foldtrust as ft
import matplotlib.pyplot as plt

print(f"FoldTrust version: {ft.__version__}")
```

## Load sequence

```python
# Load SARS-CoV-2 frameshift stimulatory element
fd = ft.io.read_case_directory("../../data/cases/sars2-fse")

print(fd)
print(f"\nSequence length: {fd.n_obs} nt")
print(f"Disease: {fd.uns.get('disease', 'Unknown')}")
```

## Run analysis pipeline

```python
# Run complete pipeline: MFE + ensemble + tiers + unpaired probs
fd = ft.tl.run_pipeline(fd, temperature=37.0)

print("Analysis complete!")
print(f"\nMFE structure: {fd.structures['mfe'][:60]}...")
print(f"MFE energy: {fd.uns['mfe_energy']:.2f} kcal/mol")
print(f"\nTier distribution:")
print(fd.obs['tier'].value_counts())
```

## Visualize results

```python
# Arc plot showing structure colored by tier
ft.pl.arc_plot(fd, color_by_tier=True, figsize=(14, 4), show=True)
```

```python
# Pair probability heatmap
ft.pl.heatmap(fd, figsize=(8, 8))
```

```python
# Tier distribution
ft.pl.tier_distribution(fd, figsize=(6, 4))
```

## Access data

```python
# Per-nucleotide annotations are in fd.obs DataFrame
print("Per-nucleotide data:")
print(fd.obs.head(10))
```

```python
# Pairwise data (pair probability matrix) in fd.obsp
print(f"Pair probability matrix shape: {fd.obsp['pair_probs'].shape}")
print(f"Max pair probability: {fd.obsp['pair_probs'].max():.3f}")
```

```python
# Metadata in fd.uns
print("Metadata:")
for key in ['mfe_energy', 'n_firm', 'n_soft', 'n_floppy']:
    print(f"  {key}: {fd.uns.get(key, 'N/A')}")
```

## Save to disk

```python
# Save to HDF5
fd.write_h5("/tmp/sars2_fse.h5")

# Load back
fd2 = ft.FoldData.read_h5("/tmp/sars2_fse.h5")
print(f"Loaded from disk: {fd2.name}, {fd2.n_obs} nt")
```

## Batch analysis

```python
# Load all disease cases
collection = ft.io.read_case_batch("../../data/cases")
print(f"Loaded {len(collection)} cases")

# Run pipeline on all cases
collection = ft.tl.run_pipeline_batch(collection, temperature=37.0)
```

```python
# Plot summary across all cases
ft.pl.disease_summary(collection, figsize=(12, 6))
```
